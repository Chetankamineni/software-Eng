```python
t = 30
h = 50
wind = 15
w = 0.5 * (t ** 2) - (0.2 * h) + (0.1 * wind) - 15
if w > 10:
    print("Sunny")
elif 0 <= w <= 10:
    print("Windy")
elif -10 <= w < 0:
    print("Cloudy")
else:
    print("Rainy")
```

```python
t = int(input("Enter Temperature: "))
h = int(input("Enter Humidity: "))
wind = int(input("Enter Wind Speed: "))

w = 0.5 * (t ** 2) - (0.2 * h) + (0.1 * wind) - 15

if w > 10:
    print("Sunny")
elif 0 <= w <= 10:
    print("Windy")
elif -10 <= w < 0:
    print("Cloudy")
else:
    print("Rainy")
```
